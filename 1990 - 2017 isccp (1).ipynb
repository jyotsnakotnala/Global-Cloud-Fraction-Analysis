```python
import matplotlib.pyplot as plt

# ======================================================
# Select common period
# ======================================================
obs = isccp_series["1990":"2017"]
ukesm_plot = ukesm["1990":"2017"]
ipsl_plot = ipsl["1990":"2017"]
mri_plot = mri["1990":"2017"]

# ======================================================
# Calculate climatological mean (1991–2017)
# ======================================================
obs_clim = obs["1991":"2017"].mean()
ukesm_clim = ukesm_plot["1991":"2017"].mean()
ipsl_clim = ipsl_plot["1991":"2017"].mean()
mri_clim = mri_plot["1991":"2017"].mean()

# ======================================================
# Compute anomalies
# ======================================================
obs_anom = obs - obs_clim
ukesm_anom = ukesm_plot - ukesm_clim
ipsl_anom = ipsl_plot - ipsl_clim
mri_anom = mri_plot - mri_clim

# ======================================================
# Apply 12-month running mean
# ======================================================
obs_anom = obs_anom.rolling(12, center=True).mean()
ukesm_anom = ukesm_anom.rolling(12, center=True).mean()
ipsl_anom = ipsl_anom.rolling(12, center=True).mean()
mri_anom = mri_anom.rolling(12, center=True).mean()

# ======================================================
# Plot
# ======================================================
plt.style.use("ggplot")

fig, ax = plt.subplots(figsize=(16,8))

fig.patch.set_facecolor("white")
ax.set_facecolor("#f8f9fa")

# ISCCP
ax.plot(
    obs_anom.index,
    obs_anom,
    color="black",
    linewidth=3,
    label="ISCCP"
)

# UKESM
ax.plot(
    ukesm_anom.index,
    ukesm_anom,
    color="royalblue",
    linewidth=2.2,
    label="UKESM1-0-LL"
)

# IPSL
ax.plot(
    ipsl_anom.index,
    ipsl_anom,
    color="darkorange",
    linewidth=2.2,
    label="IPSL-CM6A-LR"
)

# MRI
ax.plot(
    mri_anom.index,
    mri_anom,
    color="forestgreen",
    linewidth=2.2,
    label="MRI-ESM2-0"
)

# Zero anomaly line
ax.axhline(
    0,
    color="gray",
    linestyle="--",
    linewidth=1.5
)

# Labels
ax.set_title(
    "Global Mean Cloud Fraction Anomalies (1990–2017)\n"
    "ISCCP Observations and CMIP6 Model Simulations",
    fontsize=20,
    fontweight="bold",
    pad=18
)

ax.set_xlabel(
    "Year",
    fontsize=15,
    fontweight="bold"
)

ax.set_ylabel(
    "Cloud Fraction Anomaly (%)",
    fontsize=15,
    fontweight="bold"
)

# Grid
ax.grid(
    True,
    linestyle="--",
    linewidth=0.7,
    alpha=0.4
)

# Legend
legend = ax.legend(
    fontsize=12,
    frameon=True,
    loc="upper right"
)

legend.get_frame().set_facecolor("white")
legend.get_frame().set_edgecolor("gray")

# Axis styling
ax.tick_params(axis='both', labelsize=12)

for spine in ax.spines.values():
    spine.set_linewidth(1.2)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# =====================================================
# Select study period
# =====================================================
obs = isccp_series["1990":"2017"]
ukesm_plot = ukesm["1990":"2017"]
ipsl_plot = ipsl["1990":"2017"]
mri_plot = mri["1990":"2017"]

# =====================================================
# Monthly climatology
# =====================================================
obs_clim = obs.groupby(obs.index.month).mean()
ukesm_clim = ukesm_plot.groupby(ukesm_plot.index.month).mean()
ipsl_clim = ipsl_plot.groupby(ipsl_plot.index.month).mean()
mri_clim = mri_plot.groupby(mri_plot.index.month).mean()

# =====================================================
# Month labels
# =====================================================
months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

# =====================================================
# Professional Plot
# =====================================================
plt.style.use("ggplot")

fig, ax = plt.subplots(figsize=(12,7))

# Background
fig.patch.set_facecolor("white")
ax.set_facecolor("#f8f9fa")

# =====================================================
# Plot lines
# =====================================================
ax.plot(
    months,
    obs_clim,
    color="black",
    linewidth=3,
    marker="o",
    markersize=7,
    label="ISCCP"
)

ax.plot(
    months,
    ukesm_clim,
    color="royalblue",
    linewidth=2.5,
    marker="s",
    markersize=6,
    label="UKESM1-0-LL"
)

ax.plot(
    months,
    ipsl_clim,
    color="darkorange",
    linewidth=2.5,
    marker="^",
    markersize=6,
    label="IPSL-CM6A-LR"
)

ax.plot(
    months,
    mri_clim,
    color="forestgreen",
    linewidth=2.5,
    marker="D",
    markersize=6,
    label="MRI-ESM2-0"
)

# =====================================================
# Title & Labels
# =====================================================
ax.set_title(
    "Climatological Seasonal Cycle of Cloud Fraction (1990–2017)",
    fontsize=18,
    fontweight="bold",
    pad=15
)

ax.set_xlabel(
    "Month",
    fontsize=14,
    fontweight="bold"
)

ax.set_ylabel(
    "Cloud Fraction (%)",
    fontsize=14,
    fontweight="bold"
)

# =====================================================
# Grid
# =====================================================
ax.grid(
    True,
    linestyle="--",
    linewidth=0.6,
    alpha=0.4
)

# =====================================================
# Legend
# =====================================================
legend = ax.legend(
    fontsize=12,
    loc="best",
    frameon=True
)

legend.get_frame().set_facecolor("white")
legend.get_frame().set_edgecolor("gray")
legend.get_frame().set_alpha(0.95)

# =====================================================
# Axis Styling
# =====================================================
ax.tick_params(axis="both", labelsize=12)

for spine in ax.spines.values():
    spine.set_linewidth(1.2)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# ==========================================================
# Restrict study period
# ==========================================================

obs = obs_harmonized["1990":"2025"]
ukesm_plot = ukesm["1990":"2025"]
ipsl_plot = ipsl["1990":"2025"]
mri_plot = mri["1990":"2025"]

# ==========================================================
# Monthly climatology
# ==========================================================

obs_clim = obs.groupby(obs.index.month).mean()
ukesm_clim = ukesm_plot.groupby(ukesm_plot.index.month).mean()
ipsl_clim = ipsl_plot.groupby(ipsl_plot.index.month).mean()
mri_clim = mri_plot.groupby(mri_plot.index.month).mean()

# ==========================================================
# Calculate anomalies
# ==========================================================

obs_anom = obs.copy()
ukesm_anom = ukesm_plot.copy()
ipsl_anom = ipsl_plot.copy()
mri_anom = mri_plot.copy()

for m in range(1,13):

    obs_anom.loc[obs.index.month == m] -= obs_clim[m]

    ukesm_anom.loc[ukesm_plot.index.month == m] -= ukesm_clim[m]

    ipsl_anom.loc[ipsl_plot.index.month == m] -= ipsl_clim[m]

    mri_anom.loc[mri_plot.index.month == m] -= mri_clim[m]

# ==========================================================
# 12-month running mean
# ==========================================================

obs_sm = obs_anom.rolling(12, center=True).mean()

ukesm_sm = ukesm_anom.rolling(12, center=True).mean()

ipsl_sm = ipsl_anom.rolling(12, center=True).mean()

mri_sm = mri_anom.rolling(12, center=True).mean()

# ==========================================================
# Plot
# ==========================================================

plt.style.use("seaborn-v0_8-whitegrid")

plt.figure(figsize=(16,7))

plt.plot(
    obs_sm,
    color="black",
    linewidth=3,
    label="Harmonized ISCCP + CERES"
)

plt.plot(
    ukesm_sm,
    color="royalblue",
    linewidth=2,
    label="UKESM1-0-LL"
)

plt.plot(
    ipsl_sm,
    color="darkorange",
    linewidth=2,
    label="IPSL-CM6A-LR"
)

plt.plot(
    mri_sm,
    color="forestgreen",
    linewidth=2,
    label="MRI-ESM2-0"
)

plt.axhline(
    0,
    color="gray",
    linestyle="--",
    linewidth=1.5
)

plt.title(
    "Cloud Fraction Anomalies (1990–2025)",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(
    "Year",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Cloud Fraction Anomaly (%)",
    fontsize=13,
    fontweight="bold"
)

plt.legend(
    frameon=False,
    fontsize=11
)

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

```python
import numpy as np
import pandas as pd

# ==========================================================
# 1. Tropical ISCCP (30°S–30°N)
# ==========================================================

# Tropical mask using equal-area cell latitude
mask = np.abs(isccp["eqlat"].isel(time=0)) <= 30

# Mean over tropical equal-area cells
trop_isccp = (
    isccp["cldamt"]
    .where(mask)
    .mean(dim="eqcell")
    .compute()
    .to_series()
)

# Create monthly datetime index (1990-01 to 2017-06)
trop_isccp.index = pd.date_range(
    start="1990-01-01",
    periods=len(trop_isccp),
    freq="MS"
)

# ==========================================================
# 2. Tropical CERES (30°S–30°N)
# ==========================================================

weights = np.cos(np.deg2rad(ceres.lat))

trop_ceres = (
    ceres["cldarea_total_mon"]
    .sel(lat=slice(-30,30))
    .weighted(weights.sel(lat=slice(-30,30)))
    .mean(("lat","lon"))
    .to_series()
)

# ==========================================================
# 3. Compute Mean Bias (2001–2017)
# ==========================================================

overlap = pd.DataFrame({
    "ISCCP": trop_isccp["2001":"2017"],
    "CERES": trop_ceres["2001":"2017"]
}).dropna()

bias = (overlap["CERES"] - overlap["ISCCP"]).mean()

print("Mean Tropical Bias =", bias)

# ==========================================================
# 4. Harmonize ISCCP
# ==========================================================

trop_isccp_corrected = trop_isccp + bias

# ==========================================================
# 5. Final Tropical Observation (1990–2025)
# ==========================================================

trop_obs = pd.concat([
    trop_isccp_corrected["1990":"2017"],
    trop_ceres["2018":"2025"]
]).sort_index()

print(trop_obs.head())
print(trop_obs.tail())
```
