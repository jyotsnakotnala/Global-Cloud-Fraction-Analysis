```python
import xarray as xr
import pandas as pd
import glob
import re

files = sorted(glob.glob("ISCCP.HGM*.nc"))

results = []

for f in files:

    ds = xr.open_dataset(f)

    match = re.search(r'GLOBAL\.(\d{4})\.(\d{2})', f)
    year = int(match.group(1))
    month = int(match.group(2))

    cldamt = ds["cldamt"].mean().item()
    tau    = ds["tau"].mean().item()
    wp     = ds["wp"].mean().item()
    pc     = ds["pc"].mean().item()
    tc     = ds["tc"].mean().item()

    rs         = ds["rs"].mean().item()
    rs_clrsky  = ds["rs_clrsky"].mean().item()

    ts         = ds["ts"].mean().item()
    ts_clrsky  = ds["ts_clrsky"].mean().item()

    sw_cre = rs_clrsky - rs
    lw_cre = ts - ts_clrsky
    net_cre = sw_cre + lw_cre

    results.append({
        "date": pd.Timestamp(year, month, 1),
        "cldamt": cldamt,
        "tau": tau,
        "wp": wp,
        "pc": pc,
        "tc": tc,
        "SW_CRE": sw_cre,
        "LW_CRE": lw_cre,
        "NET_CRE": net_cre
    })

df = pd.DataFrame(results)
df = df.sort_values("date")

df
```

```python
import xarray as xr
import pandas as pd
import glob
import re
import matplotlib.pyplot as plt

# Read all NetCDF files
files = sorted(glob.glob("*.nc"))

results = []

for f in files:

    ds = xr.open_dataset(f)

    # Extract year and month from filename
    match = re.search(r'GLOBAL\.(\d{4})\.(\d{2})', f)
    year = int(match.group(1))
    month = int(match.group(2))

    # Global means
    cldamt = ds["cldamt"].mean(skipna=True).item()
    tau = ds["tau"].mean(skipna=True).item()
    wp = ds["wp"].mean(skipna=True).item()
    pc = ds["pc"].mean(skipna=True).item()
    tc = ds["tc"].mean(skipna=True).item()

    rs = ds["rs"].mean(skipna=True).item()
    rs_clrsky = ds["rs_clrsky"].mean(skipna=True).item()

    ts = ds["ts"].mean(skipna=True).item()
    ts_clrsky = ds["ts_clrsky"].mean(skipna=True).item()

    # Cloud Radiative Effects
    sw_cre = rs_clrsky - rs
    lw_cre = ts - ts_clrsky
    net_cre = sw_cre + lw_cre

    results.append({
        "Date": pd.Timestamp(year, month, 1),
        "CloudAmount": cldamt,
        "OpticalDepth": tau,
        "WaterPath": wp,
        "CloudTopPressure": pc,
        "CloudTopTemperature": tc,
        "SW_CRE": sw_cre,
        "LW_CRE": lw_cre,
        "NET_CRE": net_cre
    })

# Create dataframe
df = pd.DataFrame(results)
df = df.sort_values("Date")

# Save CSV
df.to_csv("ISCCP_HGM_1990_2000_CRE.csv", index=False)

# Variables to plot
variables = [
    "CloudAmount",
    "OpticalDepth",
    "WaterPath",
    "CloudTopPressure",
    "CloudTopTemperature",
    "SW_CRE",
    "LW_CRE",
    "NET_CRE"
]

# Generate and save plots
for var in variables:

    plt.figure(figsize=(12,5))
    plt.plot(df["Date"], df[var])
    plt.title(f"{var} (1990-2000)")
    plt.xlabel("Year")
    plt.ylabel(var)
    plt.grid(True)

    plt.savefig(f"{var}_1990_2000.png",
                dpi=300,
                bbox_inches="tight")

    plt.show()

print("Analysis Complete")
print("CSV file saved: ISCCP_HGM_1990_2000_CRE.csv")
print("All graphs saved as PNG files")
```

```python
import os
import glob
import re
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt

# Force plots to embed cleanly into Jupyter outputs
%matplotlib inline

# 1. Path to your ISCCP files
isccp_dir = "/Users/jyotikotnala/ISCCP_HGM/HGM_1990_2000"
isccp_files = sorted(glob.glob(os.path.join(isccp_dir, "ISCCP.HGM*.nc")))

# 2. Extract data from 1990 to 2000
results = []
for f in isccp_files:
    try:
        ds = xr.open_dataset(f)
        match = re.search(r'GLOBAL\.(\d{4})\.(\d{2})', os.path.basename(f))
        if match:
            year = int(match.group(1))
            month = int(match.group(2))
            
            # Extract means for physical cloud variables (excluding cloud fraction)
            results.append({
                "Date": pd.Timestamp(year, month, 1),
                "OpticalDepth": ds["tau"].mean(skipna=True).item(),
                "WaterPath": ds["wp"].mean(skipna=True).item(),
                "CloudTopPressure": ds["pc"].mean(skipna=True).item(),
                "CloudTopTemperature": ds["tc"].mean(skipna=True).item()
            })
    except Exception as e:
        print(f"Error reading {f}: {e}")

# Create DataFrame filtered to your exact timeline
df = pd.DataFrame(results).sort_values("Date")
df = df[(df["Date"] >= "1990-01-01") & (df["Date"] <= "2000-12-31")]

# 3. Plot a 2x2 grid of the variables directly in the output
fig, axes = plt.subplots(2, 2, figsize=(15, 10), sharex=True)
fig.suptitle("ISCCP Cloud Properties Timeline (1990 - 2000)", fontsize=16, fontweight='bold', y=0.95)

# Plot 1: Optical Depth
axes[0, 0].plot(df["Date"], df["OpticalDepth"], color="royalblue", linewidth=1.5)
axes[0, 0].set_title("Mean Cloud Optical Depth ($\\tau$)", fontsize=12)
axes[0, 0].grid(True, linestyle="--", alpha=0.5)
axes[0, 0].set_ylabel("Count / Thickness")

# Plot 2: Water Path
axes[0, 1].plot(df["Date"], df["WaterPath"], color="forestgreen", linewidth=1.5)
axes[0, 1].set_title("Mean Cloud Water Path ($WP$)", fontsize=12)
axes[0, 1].grid(True, linestyle="--", alpha=0.5)
axes[0, 1].set_ylabel("Count / g/m² equivalent")

# Plot 3: Cloud Top Pressure
axes[1, 0].plot(df["Date"], df["CloudTopPressure"], color="darkorchid", linewidth=1.5)
axes[1, 0].set_title("Mean Cloud Top Pressure ($P_c$)", fontsize=12)
axes[1, 0].grid(True, linestyle="--", alpha=0.5)
axes[1, 0].set_ylabel("Pressure (mb-equivalent count)")
axes[1, 0].set_xlabel("Year")

# Plot 4: Cloud Top Temperature
axes[1, 1].plot(df["Date"], df["CloudTopTemperature"], color="crimson", linewidth=1.5)
axes[1, 1].set_title("Mean Cloud Top Temperature ($T_c$)", fontsize=12)
axes[1, 1].grid(True, linestyle="--", alpha=0.5)
axes[1, 1].set_ylabel("Temperature (K-equivalent count)")
axes[1, 1].set_xlabel("Year")

plt.tight_layout(rect=[0, 0, 1, 0.92])

# Render the graph directly in your cell output
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Generate dummy dates for ISCCP (1990-2000) and CERES (2001-2025)
dates_isccp = pd.date_range(start='1990-01-01', end='2000-12-01', freq='MS')
dates_ceres = pd.date_range(start='2001-01-01', end='2025-12-01', freq='MS')

# Simulate SW_CRE, LW_CRE, NET_CRE for ISCCP (1990-2000)
np.random.seed(42)
isccp_sw = -50 + np.sin(np.linspace(0, 11 * 2 * np.pi, len(dates_isccp))) + np.random.normal(0, 1, len(dates_isccp))
isccp_lw = 25 + 0.5 * np.cos(np.linspace(0, 11 * 2 * np.pi, len(dates_isccp))) + np.random.normal(0, 0.5, len(dates_isccp))
isccp_net = isccp_sw + isccp_lw

# Simulate CERES (2001-2025)
ceres_sw = -48 + 1.2 * np.sin(np.linspace(0, 25 * 2 * np.pi, len(dates_ceres))) + np.random.normal(0, 0.8, len(dates_ceres))
ceres_lw = 26 + 0.4 * np.cos(np.linspace(0, 25 * 2 * np.pi, len(dates_ceres))) + np.random.normal(0, 0.4, len(dates_ceres))
ceres_net = ceres_sw + ceres_lw

fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)

# Shortwave
axes[0].plot(dates_isccp, isccp_sw, label='ISCCP (1990-2000)', color='teal')
axes[0].plot(dates_ceres, ceres_sw, label='CERES (2001-2025)', color='darkorange')
axes[0].set_title('Shortwave Cloud Radiative Effect (SW_CRE)')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend()

# Longwave
axes[1].plot(dates_isccp, isccp_lw, color='teal')
axes[1].plot(dates_ceres, ceres_lw, color='darkorange')
axes[1].set_title('Longwave Cloud Radiative Effect (LW_CRE)')
axes[1].grid(True, linestyle='--', alpha=0.5)

# Net
axes[2].plot(dates_isccp, isccp_net, color='teal')
axes[2].plot(dates_ceres, ceres_net, color='darkorange')
axes[2].set_title('Net Cloud Radiative Effect (NET_CRE)')
axes[2].grid(True, linestyle='--', alpha=0.5)
axes[2].set_xlabel('Year')

plt.tight_layout()
plt.savefig('combined_isccp_ceres_cre_timeline.png', dpi=300)
print("Saved CRE plot")
```
